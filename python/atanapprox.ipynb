```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def coeffs(degree, points=1000):
    # 1. Setup the data
    x = np.linspace(-1, 2, points)
    y = np.arctan(x)

    # 2. Perform the polynomial fit
    # returns coefficients in descending order: [p_n, p_n-1, ..., p_0]
    coeffs = np.polyfit(x, y, degree)

    # 3. Create a polynomial object for easy evaluation
    poly_func = np.poly1d(coeffs)

    # 4. Evaluation and Error Analysis
    x_fine = np.linspace(0, 1, 200)
    y_poly = poly_func(x_fine)
    y_actual = np.arctan(x_fine)
    error = y_actual - y_poly

    print(f"Polynomial Coefficients (Degree {degree}):")
    print(coeffs)

    # --- Visualization ---
    plt.figure(figsize=(10, 5))

    # Plot Comparison
    plt.subplot(1, 2, 1)
    plt.plot(x_fine, y_actual, 'k--', label='Actual arctan(x)', alpha=0.6)
    plt.plot(x_fine, y_poly, 'r', label=f'Degree {degree} Fit')
    plt.title(f'Polyfit of arctan(x) (n={degree})')
    plt.legend()
    plt.grid(True)

    # Plot Residual Error
    plt.subplot(1, 2, 2)
    plt.plot(x_fine, error, color='purple')
    plt.title('Residual Error (Actual - Fit)')
    plt.axhline(0, color='black', lw=1)
    plt.grid(True)

    plt.tight_layout()
    plt.show()
    return coeffs
```

```python
coeffs(11)
```
